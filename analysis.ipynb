```python
import matplotlib.pyplot as plt
import seaborn as sns

import numpy as np
```

```python
loaded = np.load("sim_sd_10q.npy")

data = loaded[:-1]
ideal = loaded[-1]
```

```python
ideal
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
        
ax.plot(data[:, 0], color="blue", label="Z", linewidth=0.5)
ax.plot(data[:, 1], color="green", label="Y", linewidth=0.5)
ax.plot(data[:, 2], color="orange", label="X", linewidth=0.5)

ax.axhline(y=ideal[0], color="red", linestyle="--", linewidth=1)
ax.axhline(y=ideal[1], color="red", linestyle="--", linewidth=1)
ax.axhline(y=ideal[2], color="red", linestyle="--", linewidth=1)

ax.set_xlabel("Номер симуляции")
ax.set_ylabel("Состояние измеренного кубита")
ax.set_title("Результаты телепортации на SD симуляторе (10 кубитов)")
ax.legend()
ax.grid(True)

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

ax.bar(x="x", height=data[:, 0].std())
ax.bar(x="y", height=data[:, 1].std())
ax.bar(x="z", height=data[:, 2].std())

ax.set_xlabel("Ось")
ax.set_title("Стандартное отклонение на ось")
ax.legend()

plt.tight_layout()
plt.show()
```
